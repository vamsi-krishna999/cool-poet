```python
#create a student class with student name and studnt marks in 5 different subjects and give me the sum of all the subjects using methofs


class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def totalmarks(self):
    return sum(self.marks)

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())



##############alternatively

class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total


s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
```

```python
#Static methods: without using self
class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total

  @staticmethod
  def sayhello():
    print("heymate")

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
print(s1.sayhello())
```

```python
#single inheritance

class Animal:
  def __init__(self,name):
    self.name=name

  def speak(self):
    return f"{self.name} makes a sound"

  def eat(self):
    print("eating")

class dog(Animal):
  def speak(self):
    return f"{self.name} barks"

dog1=dog("tomy")

print(dog1.speak())
print(dog1.eat())
```

```python
#create multiple inheritance with methods in it


class aeroplace():
  def __init__(self,engine,fname):
    self.engine=engine
    self.fname=fname

  def takeoff(self):
    return f"{self.fname}  has takenoff"

  @staticmethod
  def touchdown():
    print('plane will land with autopilot')


class hanger():
  def __init__(self,name):
    self.name=name

  def hold(self):
    return f"{self.name} is  hanger"

class jetplane(aeroplace,hanger):
  def touchdown():
    print('plane will land with autopilot')

f16_raptor=jetplane('rollroyce','raptor','nevada')
print(f16_raptor.takeoff())
```

```python
class aeroplace:
    def __init__(self, engine, fname):
        self.engine = engine
        self.fname = fname

    def takeoff(self):
        return f"{self.fname} has taken off"

    @staticmethod
    def touchdown():
        print("plane will land with autopilot")


class hanger:
    def __init__(self, name):
        self.name = name

    def hold(self):
        return f"{self.name} is a hangar"


class jetplane(aeroplace, hanger):
    def __init__(self, engine, fname, name):
        aeroplace.__init__(self, engine, fname)
        hanger.__init__(self, name)


f16_raptor = jetplane("rollroyce", "raptor", "nevada")

print(f16_raptor.takeoff())
print(f16_raptor.hold())
f16_raptor.touchdown()
```

#Super Method
it is used to access methods of parent class in the child class

```python
class car:
  def __init__(self,type):
    self.type=type

  def start(self):
    print("starting the car")

class toyotacar(car):
  def __init__(self,name,type):    #we need to pass type to class car
    self.name=name
    super().__init__(type)
    super().start()
x=toyotacar("corolla","sedan")
print(x.name)
print(x.type)
```

#Encapsulation

wrapping data and methods as a single unit


```python
#public access modifier: methods and attributes can be accessed everwhere in all classes
#private access modifier : methods and attributes are accessible only within the class where they are defined. ('__')
#protected access modifier: methods and attributes are accessible within the class and its subclass. (use single underscore)

```

```python
class car:
  def __init__(self,make,model):
    self.__make=make
    self.__model=model

  def __display(self): #two underscores to make it private
    return f"{self.__make}{self.__model}"

  def get_make(self): #setter-- Public access modifie used to display private methods-
    return self.__display()

car1=car("Honda","Civic")

print(car1.get_make())
```

```python
#protected access modifier:


class car:
  def __init__(self,make,model):
    self._make=make
    self._model=model

  def _display(self):
    return f"{self._make}{self._model}"

class electriccar(car):
  def battery_info(self):
    return f"{self._make}{self._model} has a battery"

car1=electriccar("Honda","Civic")
print(car1._make)
print(car1._model)
print(car1._display())


```

```python
#METHOD OVERRIDING

class Animal:
  def __init__(self,name):
    self.name=name

  def kill(self):
    return f"{self.name} kill to eat"

class Predator(Animal):
  def kill(self):  #Same Method in parent class is overriden
    return f"{self.name} ki for fun"

pred=Predator("tiger")
print(pred.kill())
```

```python
# DUCK TYPING
#type of the object is determined by its behavior and not from its inheritance from a specific class


class duck:
  def speak(self):
    print("quack quack")

class dog:
  def speak(self):
    print("bow bow")

def make_sound(animal): # only checks if dog and duck has speak
  animal.speak()

duck1=duck()
dog1=dog()

make_sound(duck1)
make_sound(dog1)
```

```python
#ABSTRACTION : hiding the implementation details

from abc import ABC,abstractmethod


class Alpha(ABC): #abstract class, methods here should not have a body
  @abstractmethod #just a blueprint should not have any values
  def prey(self):
    pass

class Sigma(Alpha):
  def prey(self):
    print("start hunt")

Sigma1=Sigma()
print(Sigma1.prey())
```
