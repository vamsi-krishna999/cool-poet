```python
#methods are the functions that are defined within a class -- they define the behavior of the objects


class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def display(self):
    print("welcome student",self.name)
    print("your marks are ",self.marks)
    return 'hello'

s1=student("vamsi",100)
s2=student("sahi",100)

print(s1.display())
```

```python
#create a student class with student name and studnt marks in 5 different subjects and give me the sum of all the subjects using methofs


class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def totalmarks(self):
    return sum(self.marks)

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())



##############alternatively

class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total


s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
```

```python
#Static methods: without using self
class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total

  @staticmethod
  def sayhello():
    print("heymate")

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
print(s1.sayhello())
```
