```python
#Create a dataframe from a dictionary

data ={"name":['vamsi','Krishna','Kal'],
       'age':[23,24,25],
       'city':['vizag','bangalore','sydney']}

df=pd.DataFrame(data)
print(df)
```

```python
#Access the data - iloc


print(df.iloc[0,0]) #(row,column)
print("--------------")
print(df.iloc[2]) #full row
print("--------------")
print(df.iloc[0:2]) #first two rows

print("--------------")
print("--------------")
print(df.iloc[:,0:2]) #all rows but only first two columns
```

```python
#loc - without numbers actual column names

print (df.loc[2,'age']) #2nd row and afge
print("--------------")
print(df.loc[2]) #full row
print("--------------")

print(df.loc[0:2]) #first two rows
print("--------------")

print(df.loc[:,'name':'city']) #all rows but only first two columns



```

#MERGE




like joins


```python
#merge

import pandas as pd

Student_data={"student_id":[101,102,103,104]
,
"name":['vamsi','sahithi','farhana','sravani']
}

Marks_data={"student_id":[101,102,103,104],
            "marks":[80,90,100,65]}
students=pd.DataFrame(Student_data)
marks=pd.DataFrame(Marks_data)

students.head()
marks.head()
```

```python
#inner merger operation


result=pd.merge(students,marks,on='student_id',how='inner')
print(result)
```

```python
#Left /Joibn

leftjoin=pd.merge(students,marks,on='student_id',how='left') #everything on left table needs to come

print(leftjoin)
```

```python
#right Join

right=pd.merge(students,marks,on='student_id',how='right')
print(right)
```

```python
#full outer join
#right Join

full=pd.merge(students,marks,on='student_id',how='outer')
print(full)
```

```python
#merge with different column names
import pandas as pd

students=pd.DataFrame({"student_id":[101,102,103,104]
,
"name":['vamsi','sahithi','farhana','sravani']
}
)
marks=pd.DataFrame({"id":[101,102,103,105],
            "marks":[80,90,100,65]})

result=pd.merge(students,marks,left_on="student_id",right_on='id',how='inner')
print(result)


#Joining on multiple columns

#result=pd.merge(student,marks,on=['emp_id','year'],how='inner')
```

```python
#concatenaton

#difference between union all and union : union all givces duplicates and not the others

students1=pd.DataFrame({"student_id":[101,102,103,104]
,
"name":['vamsi','sahithi','farhana','sravani'],
                        "marks":[23,34,5,6]
})

students2=pd.DataFrame({"student_id":[101,102,103,104]
,
"name":['va','sah','fah','sra'],
                        "marks":[34,4,44,4]
})


rconcat=pd.concat([students1,students2],axis=0)
print(rconcat)
print("*"*50)
rconcat_col=pd.concat([students1,students2],axis=1)
print(rconcat_col)

##indexes need to be defined as they are same after concatenationb
```

```python
#Reset INDEX
concat_resetindex=pd.concat([students1,students2]).reset_index()
print(concat_resetindex)
```

```python
#ignore index


concat_ignoreindex=pd.concat([students1,students2],ignore_index=True)
print(concat_ignoreindex)
```
