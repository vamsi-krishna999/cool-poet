```python
#REDUCE

#Reducing the input to a singlar output
#Need to be imported from libraries
#In Python, reduce() is used when you want to take many values and combine them into one final value.
from functools import reduce # Not readily available install them using pip install and them import them to elevate the python functionality


lst=[1,2,3,4,45,5,5,6]

sum=reduce(lambda x,y:x+y,lst)
print(sum)
```

```python
#decorator:

#Dry - Dont repeat yourself, we dont have to repeat it
#Decoration around your function, enhancing
#Put it inside a wrapper

def my_decorator1(func):
  def wrapper1():
    print("before execution")
    func()
    print("after execution")
  return wrapper1
@my_decorator1
def say_hello():
  print("hello")
say_hello()
```

```python
#Decorator with login example

def login_required(func):
  def wrapper():
    print("Checking if users have access")
    logged_in=True #false

    if logged_in:
      func()
    else:
      print("access denied")
  return wrapper

@login_required
def view_profile():
  print("displaying profile")
view_profile()
```
