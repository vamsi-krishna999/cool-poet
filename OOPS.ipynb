```python
#methods are the functions that are defined within a class -- they define the behavior of the objects


class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def display(self):
    print("welcome student",self.name)
    print("your marks are ",self.marks)
    return 'hello'

s1=student("vamsi",100)
s2=student("sahi",100)

print(s1.display())
```

```python
#create a student class with student name and studnt marks in 5 different subjects and give me the sum of all the subjects using methofs


class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

  def totalmarks(self):
    return sum(self.marks)

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())



##############alternatively

class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total


s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
```

```python
#Static methods: without using self
class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
  def totalmarks(self):
    total=0
    for n in self.marks:
      total=total+n
    return total

  @staticmethod
  def sayhello():
    print("heymate")

s1=student("Vamsi",[70,60,40])
print(s1.totalmarks())
print(s1.sayhello())
```

```python
#single inheritance

class Animal:
  def __init__(self,name):
    self.name=name

  def speak(self):
    return f"{self.name} makes a sound"

  def eat(self):
    print("eating")

class dog(Animal):
  def speak(self):
    return f"{self.name} barks"

dog1=dog("tomy")

print(dog1.speak())
print(dog1.eat())
```

```python
#multiple inheritance  : Multiple parent classes but one child class


class a:
  vara='hello'

class b:
  varb='bye'

class c(a,b):
  pass

new=c()
print(new.vara)
print(new.varb)
```

```python
#create multiple inheritance with methods in it


class aeroplace():
  def __init__(self,engine,fname):
    self.engine=engine
    self.fname=fname

  def takeoff(self):
    return f"{self.fname}  has takenoff"

  @staticmethod
  def touchdown():
    print('plane will land with autopilot')


class hanger():
  def __init__(self,name):
    self.name=name

  def hold(self):
    return f"{self.name} is  hanger"

class jetplane(aeroplace,hanger):
  def touchdown():
    print('plane will land with autopilot')

f16_raptor=jetplane('rollroyce','raptor','nevada')
print(f16_raptor.takeoff())
```

```python
class aeroplace:
    def __init__(self, engine, fname):
        self.engine = engine
        self.fname = fname

    def takeoff(self):
        return f"{self.fname} has taken off"

    @staticmethod
    def touchdown():
        print("plane will land with autopilot")


class hanger:
    def __init__(self, name):
        self.name = name

    def hold(self):
        return f"{self.name} is a hangar"


class jetplane(aeroplace, hanger):
    def __init__(self, engine, fname, name):
        aeroplace.__init__(self, engine, fname)
        hanger.__init__(self, name)


f16_raptor = jetplane("rollroyce", "raptor", "nevada")

print(f16_raptor.takeoff())
print(f16_raptor.hold())
f16_raptor.touchdown()
```
