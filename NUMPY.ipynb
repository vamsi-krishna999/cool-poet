```python
#Array Reshape


arr=np.array([1,2,3,5,6,7])
array2d=arr.reshape(2,3)
print(array2d)
```

```python
#Array Flattening -- converts in a oneD array


marks=np.array([
    [1,2,3],
    [4,5,6]


 ])

flat_arry=marks.flatten()
print(flat_arry)
```

```python
#reshaping and flattening


arr=np.arange(24).reshape(2,3,4)
print(arr)

print(arr.shape)

new_arr=arr.reshape(4,6)
print(new_arr)

oned=new_arr.flatten()
print(oned)
```

```python
arr=np.arange(6)
print(arr[0])
print(arr[0:5])
print(arr[0:4:2]) # 4 is stopping element , 2 is the steps of increment
print(arr[0::2])
```

```python
y=np.arange(6).reshape(2,3)
print(y)

print(y[1,1])
print(y[0,2])
```

#Arithmatic operation

```python
a=np.array([5,43,5,6])
b=np.array([4,5,6,9])

np.add(a,b)  #ADD

np.subtract(a,b) #subtract

np.multiply(a,b)
np.divide(a,b)
np.mean(a)
np.median(a)
np.std(a)
np.sum(a)
np.min(a)
np.max(a)
np.var(a)
#has to be off same lenght, unless the array is a of a single element
```

```python
#transpose: Rows to columns and columns to rows


arr=np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr.T) #Transpose
```

```python
#concatenation
#Joining arrays together


a=np.array([
      [1,2],
      [3,4]
])

b=np.array([
      [5,6],
      [7,8]
])


c=np.concatenate((a,b),axis=0) #o is row wise concate
d=np.concatenate((a,b),axis=1) #Column wise concatenate
print(c)
print(d)
```

```python
#Stacking

a=np.array([1,2,3])
b=np.array([2,3,4])

c=np.stack((a,b),axis=0)
print(c)
```

```python
print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
#SPLIT


arr=np.array([[1,2,3],[4,5,6]])
print(arr)
newarr=np.hsplit(arr,3) #horizontally divide into 3 different array
print(newarr)

verarr=np.vsplit(arr,2)
print(verarr)
```
