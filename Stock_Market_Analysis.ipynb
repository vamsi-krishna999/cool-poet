```python
import yfinance as yf
import pandas as pd
```

```python
stocks=['AAPL','GOOGL','AMZN']
start_date='2019-01-01'
end_date='2026-01-01'
```

```python
#Calling Yahoo finance Library and downloading stocks data

stocks_data=yf.download(stocks,start=start_date,end=end_date)
```

```python
#look at the data

display(stocks_data.head())
```

```python
#import the data from the dataframe to a CSV

stocks_data.to_csv("stocksdata1.csv")
```

```python
#formatting the data



df=pd.read_csv(
    '/content/stocksdata1.csv',
    header=[0,1,2], #Takes first three rows as headers
    index_col=[0], #Takes First column as index column
    parse_dates=True,
    date_format='%Y-%m-%d'
    )

display(df.head())
```

```python
#Missing values - FFILL (inplace = True used to fill the missing values with the previous physical values )

df.ffill(inplace=True)
display(df)
```

```python
#Remove all the rows with null values
df.dropna(inplace=True)
display(df)
```

```python
#if the primary column has null values and the non mandatory fields have values


df.dropna(subset='columnx',inplace=True)
```

```python
#date column has been set as index here , we are filtering only 2023 records here


start_date='2023-01-01'
end_date='2023-12-31'

filtered_df=df[(df.index>=start_date) & (df.index<=end_date)]

display(filtered_df.head())
```

```python
#Percentage change
#to check daily returns(share price percentage change 102-100/100)


df=pd.DataFrame({
        'date':pd.date_range(start='2024-01-01',periods=5,freq='D'),
        'price':[100,102,101,108,110]
    })

df['date']=pd.to_datetime(df['date'])
df['daily_return']=df['price'].pct_change() #Percentage change

display(df)
```

```python
#Moving average- provide average of first n , then n-1 , n-2 and so on

#To track longterm trends eg for3000 days or so on

df=pd.DataFrame({
        'date':pd.date_range(start='2024-01-01',periods=30,freq='D'),
        'close':[i+100 for i in range(30)]

        })

df['date']=pd.to_datetime(df['date'])
df['movingaverage']=df['close'].rolling(window=5).mean()
display(df)
```

```python
#Understand Market Volatility- Standard Deviation

df=pd.DataFrame({
    'date':pd.date_range(start="2024-01-01",periods=30,freq='D'),
    'close':[i+100 for i in range(30)]

})
df['date']=pd.to_datetime(df['date'])

df['daily_return']=df['close'].pct_change()
volatility=df['daily_return'].std()
print(volatility)
```

```python
DataFrame=pd.DataFrame(
    {
        'date':pd.date_range(start='2024-01-01',periods=250,freq='D'),
        'close':[i+100+np.random.randn()*5 for i in range(250)]
    }
)

DataFrame['date']=pd.to_datetime(DataFrame['date'])

plt.figure(figsize=(12,6))
plt.plot(DataFrame['date'],DataFrame['close'],label='close price',color='violet')
plt.xlabel("date")
plt.ylabel("price")
plt.title("stock price trend over time")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from os import close
#50 Day moving average and #200 day moving average plotted

DataFrame=pd.DataFrame(
    {
        'date':pd.date_range(start='2024-01-01',periods=250,freq='D'),
        'close':[i+100+np.random.randn()*5 for i in range(250)]
    }
)

DataFrame['date']=pd.to_datetime(DataFrame['date'])

DataFrame['50_day_movingaverage']=DataFrame['close'].rolling(window=50).mean()

DataFrame['200_day_movingaverage']=DataFrame['close'].rolling(window=200).mean()

plt.figure(figsize=(12,6))
plt.plot(DataFrame['date'],DataFrame['close'],label='closing price',color='green')
plt.plot(DataFrame['date'],DataFrame['50_day_movingaverage'],label='50 day moving average price',color='blue')
plt.plot(DataFrame['date'],DataFrame['200_day_movingaverage'],label='200 day moving average price',color='black')
plt.xlabel("date")
plt.ylabel("price")
plt.legend()
plt.title("Stock price with moving average")
plt.grid(True)
plt.show()
```
