```python
#series

#one dimensional array, has indexes, can hold data of one data type


import pandas as pd


s=pd.Series([1,2,34,5,6])
print(s)
```

```python
s=pd.Series([1,2,3,4,5], index=['a','b','c','d','e']) #Can provide custom indexes
print(s)
print(s[1])
print(s[:3])
print(s[3:])
print(s[1:4:])
print(s[['a','c']])
```

#DataFrame

```python
#heteregenous - Multiple datatypes/ size- mutable  / Can be removed (changed)
#Two dimensional , similar to spreadsheet, tabular etc
```

```python
#Create a dataframe from a dictionary

data ={"name":['vamsi','Krishna','Kal'],
       'age':[23,24,25],
       'city':['vizag','bangalore','sydney']}

df=pd.DataFrame(data)
print(df)
```

```python
#Access the data - iloc


print(df.iloc[0,0]) #(row,column)
print("--------------")
print(df.iloc[2]) #full row
print("--------------")
print(df.iloc[0:2]) #first two rows

print("--------------")
print("--------------")
print(df.iloc[:,0:2]) #all rows but only first two columns
```

```python
#loc - without numbers actual column names

print (df.loc[2,'age']) #2nd row and afge
print("--------------")
print(df.loc[2]) #full row
print("--------------")

print(df.loc[0:2]) #first two rows
print("--------------")

print(df.loc[:,'name':'city']) #all rows but only first two columns



```

```python
data ={"name":['vamsi','Krishna','Kal','x','y','z'],
       'age':[23,24,25,30,31,32],
       'salary':[4000,23000,300000,222,3333,4444]}

df=pd.DataFrame(data)

print(df)
print("--------------")

df.head()#top 5 values
print("--------------")

df.tail()#last 5 values
```

```python
df.info() #info and data type of dataframe
```

```python
df.describe() #descriptibve stats
```

```python
#filtering the data

old=df[df['age']>30]
print(old)
```

```python
#Multiple filtering



df[(df['age']>30) & (df['salary']>3000)] #AND
print('-----------')
df[(df['age']>30) | (df['salary']>3000)] #OR
```
